This notebook performs leiden clustering and cell type annotation based on differentially expressed genes and marker gene expression.

Input: .h5ad of preprocessed Dumitru et al. dataset in ../data/dumitru/single_nucleus/h5ad/dumitru_all_donors_pp.h5ad  
Output: .h5ad of Dumitru et al. dataset with cell type annotations in ../data/dumitru/single_nucleus/h5ad/dumitru_all_donors_annotated.h5ad

```python
import scanpy as sc
sc.settings.set_figure_params(dpi=300)
```

```python
adata = sc.read_h5ad('../data/dumitru/single_nucleus/h5ad/dumitru_all_donors_pp.h5ad')
adata
```

```python
sc.tl.leiden(adata, resolution=0.2)
sc.pl.umap(adata, color=['leiden'], legend_loc='on data')
```

```python
sc.tl.rank_genes_groups(adata, groupby='leiden')
sc.pl.rank_genes_groups(adata, n_genes=25)

for cluster in adata.obs['leiden'].cat.categories:
    genes = adata.uns['rank_genes_groups']['names'][cluster][:25]
    print(f"Cluster {cluster}: {', '.join(genes)}")
```

```python
print('Granule neuron markers')
sc.pl.umap(adata, color=['NEUROD1', 'RBFOX3', 'PROX1', 'NTF3', 'DCX', 'ST8SIA2', 'GLRA2'], layer='counts', vmax='p99')
print('Excitatory neuron markers')
sc.pl.umap(adata, color=['SLC17A7', 'CAMK2A', 'NRGN', 'SATB2'], layer='counts', vmax='p99')
print('Interneuron markers')
sc.pl.umap(adata, color=['GAD1', 'GAD2', 'PVALB', 'SST', 'VIP', 'NPY', 'RELN'], layer='counts', vmax='p99')
print('Oligodendrocyte markers')
sc.pl.umap(adata, color=['MBP', 'PLP1', 'MOBP', 'SOX10', 'OLIG2'], layer='counts', vmax='p99')
print('Astrocyte markers')
sc.pl.umap(adata, color=['GFAP', 'ALDH1L1', 'SLC1A2', 'SLC1A3', 'S100B'], layer='counts', vmax='p99')
print('Endothelial cell markers')  
sc.pl.umap(adata, color=['PECAM1', 'VWF', 'CLDN5', 'SLC2A1', 'FLT1'], layer='counts', vmax='p99')
print('Ependymal cell markers')
sc.pl.umap(adata, color=['FOXJ1', 'RSPH1', 'CFAP54'], layer='counts', vmax='p99')
print('Microglia markers')
sc.pl.umap(adata, color=['P2RY12', 'TMEM119', 'AIF1', 'CX3CR1'], layer='counts', vmax='p99')
```

```python
sc.pl.umap(adata, color=['leiden'], legend_loc='on data')
```

```python
cell_type_dict = {
    '0': 'Excitatory neuron',
    '1': 'Astrocyte',
    '2': 'Interneuron',
    '3': 'Excitatory neuron',
    '4': 'Oligodendrocyte',
    '5': 'OPC',
    '6': 'Microglia',
    '7': 'Astrocyte',
    '8': 'Excitatory neuron',
    '9': 'Unknown',
    '10': 'Endothelial cell',
    '11': 'Excitatory neuron',
    '12': 'Excitatory neuron',
    '13': 'Ependymal cell',
    '14': 'Excitatory neuron',
    '15': 'OPC',
    '16': 'Excitatory neuron',
    '17': 'Excitatory neuron',
}

adata.obs['cell_type'] = adata.obs['leiden'].map(cell_type_dict).astype('category')
```

```python
sc.pl.umap(adata, color=['cell_type'], size=1)
```

```python
sc.pl.umap(adata, color=['donor'], size=1)
```

```python
adata.write_h5ad('../data/dumitru/single_nucleus/h5ad/dumitru_all_donors_global_annotations.h5ad')
```

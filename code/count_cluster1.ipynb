```python
import numpy as np
import pandas as pd
import scanpy as sc
```

```python
adata = sc.read_h5ad('../data/dumitru/single_nucleus/h5ad/dumitru_all_donors_nsc_annotations.h5ad')
adata
```

```python
# Create a summary dataframe with donor-level statistics (normalized by total cells per donor)
patient_summary = []

for donor in adata.obs['donor'].unique():
    patient_data = adata.obs[adata.obs['donor'] == donor]
    
    '''
    # Check if this donor's sample_prep is 'Not sorted'
    sample_prep = patient_data['sample_prep'].iloc[0]
    if 'enrichment' in sample_prep:
        continue  # Skip this donor
    '''
    
    # Get age for this donor (should be the same for all cells from same donor)
    age = patient_data['age'].iloc[0]
    
    # Get total number of cells for this donor
    total_cells = len(patient_data)
    
    # Count cluster 1 for this donor
    nsc1_count = (patient_data['cell_type'] == 'nsc_01').sum()
    
    patient_summary.append({
        'donor': donor,
        'age': age,
        'total_cells': total_cells,
        'nsc1_count': nsc1_count
    })

patient_df = pd.DataFrame(patient_summary)
patient_df
```

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator

fig, ax = plt.subplots(figsize=(6, 5))

ax.scatter(patient_df['age'],
           patient_df['nsc1_count'],
           c='darkblue',
           s=150,
           alpha=0.7,
           edgecolors='black',
           linewidth=0.5)

ax.set_xlabel('Donor age')
ax.set_ylabel('Number of cells')
ax.set_title('Number of cells in cluster 1 across donors', fontsize=16)

ax.yaxis.set_major_locator(MaxNLocator(integer=True))

plt.tight_layout()
plt.show()
```

```python
adult_df = patient_df[patient_df['age'] > 5]
adult_df
```

```python
# Create a summary dataframe with donor-level statistics (normalized by total cells per donor)
patient_summary = []

for donor in adata.obs['donor'].unique():
    patient_data = adata.obs[adata.obs['donor'] == donor]

    # Check if this donor's sample_prep is 'Not sorted'
    sample_prep = patient_data['sample_prep'].iloc[0]
    if 'enrichment' in sample_prep:
        continue  # Skip this donor
    
    # Get age for this donor (should be the same for all cells from same donor)
    age = patient_data['age'].iloc[0]
    
    # Get total number of cells for this donor
    total_cells = len(patient_data)
    
    # Count cluster 1 for this donor
    nsc1_count = (patient_data['cell_type'] == 'nsc_01').sum()
    
    patient_summary.append({
        'donor': donor,
        'age': age,
        'total_cells': total_cells,
        'nsc1_count': nsc1_count
    })

patient_df = pd.DataFrame(patient_summary)
patient_df
```

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator

fig, ax = plt.subplots(figsize=(6, 5))

ax.scatter(patient_df['age'],
           patient_df['nsc1_count'],
           c='darkblue',
           s=150,
           alpha=0.7,
           edgecolors='black',
           linewidth=0.5)

ax.set_xlabel('Donor age')
ax.set_ylabel('Number of cells')
ax.set_title('Number of cells in cluster 1 across donors (not sorted)', fontsize=16)

ax.yaxis.set_major_locator(MaxNLocator(integer=True))
#ax.set_ylim(-1, 20)

plt.tight_layout()
plt.show()
```

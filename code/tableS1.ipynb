This notebook creates figures from Dumitru et al.'s table S1.

Input: .xlsx in ../data/dumitru/table_S1.xlsx  
Output: figures

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('../data/dumitru/table_S1.xlsx', header=1, nrows=26)
df
```

```python
df.columns = df.columns.str.strip()
df = df[['Donor', 'Age', 'Sample Prep Protocol', 'NSC (Proliferating NSC)', 'INP (Proliferating INP)', 'NB (Proliferating NB)', 'NEUROD1+ FXYD7+ KCNQ5+ SEMA6D+ GN']]
df.columns = ['Donor', 'Age', 'Sample Prep', 'NSC', 'INP', 'NB', 'Immature GN']
df.head()
```

```python
# Remove everything after first '('
df = df.apply(lambda col: col.astype(str).str.split('(').str[0].str.strip() if col.dtype == 'object' else col)

# Convert numeric columns to integers
for col in df.columns:
    if col not in ['Donor', 'Sample Prep']:  # Skip non-numeric columns
        df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0).astype(int)
df
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

cell_types = ['NSC', 'INP', 'NB', 'Immature GN']

for i, cell_type in enumerate(cell_types):
    axes[i].scatter(df['Age'], df[cell_type], alpha=0.6, s=200)
    axes[i].set_xlabel('Age', fontsize=10)
    axes[i].set_ylabel('Number of Cells', fontsize=10)
    axes[i].set_title(cell_type, fontsize=11)

plt.tight_layout()
plt.show()
```

```python
adult_df = df[df['Age'] > 5]
adult_df
```

```python
# Plot donors > 5 years old
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

cell_types = ['NSC', 'INP', 'NB', 'Immature GN']

for i, cell_type in enumerate(cell_types):
    axes[i].scatter(adult_df['Age'], adult_df[cell_type], alpha=0.6, s=200)
    axes[i].set_xlabel('Age', fontsize=10)
    axes[i].set_ylabel('Number of Cells', fontsize=10)
    axes[i].set_title(cell_type, fontsize=11)

plt.tight_layout()
plt.show()
```

```python
adult_no_enrich_df = adult_df[~adult_df['Sample Prep'].str.contains('enrichment', case=False)]
adult_no_enrich_df
```

```python
# Plot adult donors without FACS enrichment
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
axes = axes.flatten()

cell_types = ['NSC', 'INP', 'NB', 'Immature GN']

for i, cell_type in enumerate(cell_types):
    axes[i].scatter(adult_no_enrich_df['Age'], adult_no_enrich_df[cell_type], alpha=0.6, s=200)
    axes[i].set_xlabel('Age', fontsize=10)
    axes[i].set_ylabel('Number of Cells', fontsize=10)
    axes[i].set_title(cell_type, fontsize=11)

plt.tight_layout()
plt.show()
```
